```python
S = ['0'] * len(X[0])
G = [['?'] * len(X[0])]
```

```python
def predict(S, candidate):
    for i in range(len(S)):
        if S[i] != '?' and S[i] != candidate[i]:
            return "No"
    return "Yes"
```
